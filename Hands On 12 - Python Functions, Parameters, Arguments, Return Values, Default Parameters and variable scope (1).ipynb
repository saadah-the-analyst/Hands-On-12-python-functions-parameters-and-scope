# Industry-Based Project on Function 
### Customer Data Cleaning and Profile Generator

```python
# Global Variable
COMPANY_NAME = "SmartMart"

# Function 1: Clean customer name (uses local variable 'cleaned')
def clean_name(name):
    cleaned = name.strip().title()
    return cleaned
    
# Function 2: Clean Country (uses local variable 'cleaned')
def clean_country(country):
    cleaned = country.strip().title()
    return cleaned

# Function 3: Calculate Account Status
def calculate_status(total_spent):
    if total_spent >= 100000:
        return "Premium"
    elif total_spent >= 50000:
        return "Gold"
    elif total_spent >= 20000:
        return "Silver"
    else:
        return "Standard"

# Function 4: Generate Customer Profile
def generate_customer_profile(name, country, total_spent):
    # Pass inputs through our cleaning/logic functions
    cleaned_name = clean_name(name)
    cleaned_country = clean_country(country)
    status = calculate_status(total_spent)

    # Print formatted summary output
    print("======================================================================")
    print(F"             {COMPANY_NAME.upper()} CUSTOMER PROFILE                 ")
    print("======================================================================")
    print(f"Name: {cleaned_name}")
    print(f"Country: {cleaned_country}")
    print(f"Total Spent: ₦{total_spent:,}")
    print(f"Account Status: {status}")
    print("======================================================================")

# Part A: Calling using Positional Arguments
print("--- TEST: POSITIONAL ARGUMENTS ---")
generate_customer_profile("   maria johnson ", " nigeria ", 125000)
generate_customer_profile("john smith", "ghana", 75000)
generate_customer_profile("DAVID PETER ", "kenya", 35000)

# Empty print function to leave space
print()

 # Part B: Calling using Keyword Arguments
print("--- TEST: KEYWORD ARGUMENTS ---")
generate_customer_profile(total_spent = 15000, country = " nigeria ", name = "amaka okoro")
generate_customer_profile(name = "grace williams", country = "south africa", total_spent = 85000)
```

```python
# Challenge 2 Helper Function: Calculate Loyalty Points
def calculate_loyalty_points(total_spent):
# Determine points multiplier based on spending threshold
    if total_spent >= 100000:
        multiplier = 3      # Premium tier multiplier
    elif total_spent >= 50000:
        multiplier = 2      # Gold tier multiplier
    else:
        multiplier = 1      # Silver and Standard tier multiplier

        # Calculate points earned (1 point system unit per 1,000 naria spent)
    points = int((total_spent // 1000) * multiplier)
    return points

# Main  Function : Generate Customer Profile
def generate_customer_profile(customer_id, name, country, total_spent):
    cleaned_name = clean_name(name)
    cleaned_country = clean_country(country)
    status = calculate_status(total_spent)

    # Calculate loyalty points using helper function
    loyalty_points = calculate_loyalty_points(total_spent)
    
    # print formatted customer profile summary including loyalty points
    print("======================================================================")
    print(F"             {COMPANY_NAME.upper()} CUSTOMER PROFILE                 ")
    print("======================================================================")
    print(f"Customer ID: {customer_id.strip().upper()}")
    print(f"Name: {cleaned_name}")
    print(f"Country: {cleaned_country}")
    print(f"Total Spent: ₦{total_spent:,}")
    print(f"Account Status: {status}")
    print(f"Loyalty Points: {loyalty_points}")
    print("======================================================================")
# Test Coach's Challenge 2 with Customer Data
print("--- COACH;S CHALLENGE 2: LOYALTY POINTS DISPLAY ---")
print()

generate_customer_profile("cust001", "   maria johnson ", " nigeria ", 125000)
generate_customer_profile("cust002", "john smith", "ghana", 75000)
generate_customer_profile("cust003", "DAVID PETER ", "kenya", 35000)
```

# Coach's Challenge 5 - Find the Scope Error
#### Prediction

Running this code will cause a ⁠NameError: name 'profile_title' is not defined⁠ on the final line (⁠print(profile_title)⁠).

Explanation Breakdown

 1. Where ⁠profile_title⁠ was created:

It was created inside the local scope of the ⁠generate_profile()⁠ function as a local variable.

 2. Why it works inside the function:

It works inside ⁠generate_profile()⁠ because ⁠profile_title⁠ exists within the function's local execution context. When the function runs, it can access and print any variables defined inside its own block.

 3. Why the final ⁠print()⁠ causes a problem:

Once the ⁠generate_profile()⁠ function finishes running, all of its local variables are destroyed. The final ⁠print(profile_title)⁠ is sitting in the global scope, outside the function, where ⁠profile_title⁠ does not exist. Python looks for ⁠profile_title⁠ globally, fails to find it, and throws a ⁠NameError⁠.

```python
# Coach's Challenge 5: Fixing the scope error using return
def generate_profile(name):
    profile_title = "CUSTOMER PROFILE"
    print(profile_title)
    print(name)
    return profile_title

# Returning the variable allows global scope access without throwing a NameError
profile_title = generate_profile("Maria")
print(profile_title)
```

# Bonus Challenge - Build a Mini Customer Processing System

```python

# =========================================================
# BONUS CHALLENGE: Mini Customer Processing System
# =========================================================

# Global Variable to store the company name so it can be accessed anywhere
COMPANY_NAME = "SmartBiz Logistics"

# Function to fix name formatting by removing extra spaces and capitalizing words
def clean_name(name):
    return name.strip().title()

# Function to clean country names properly 
def clean_country(country):
    return country.strip().title()

# Function to decide customer tier based on their total spending amount
def calculate_status(total_spent):
    if total_spent >= 100000:
        return "Gold"
    elif total_spent >= 50000:
        return "Silver"
    else:
        return "Bronze"

# Calculate 0.3% loyalty points from total spend
def calculate_loyalty_points(total_spent):
    return int(total_spent * 0.003)

# Main function that combines all customer details into a dictionary
# Set country = "Unknown" as default parameter if no country is passed
def generate_customer_profile(customer_id, name, total_spent, country="Unknown"):
    cleaned_name = clean_name(name)
    cleaned_country = clean_country(country) if country else "Unknown"
    status = calculate_status(total_spent)
    loyalty_points = calculate_loyalty_points(total_spent)

    return {
        "company": COMPANY_NAME, # Accessing global variable here
        "customer_id": customer_id.strip().upper(),
        "name": cleaned_name,
        "country": cleaned_country,
        "total_spent": total_spent,
        "status": status,
        "loyalty_points": loyalty_points
    }

# Dataset containing 5 raw customer records to test the system
raw_customers = [
    {"id": "cust001", "name": " maria johnson ", "country": " nigeria ", "spent": 125000},
    {"id": "cust002", "name": "john smith", "country": "ghana", "spent": 75000},
    {"id": "cust003", "name": "DAVID PETER ", "country": "", "spent": 35000}, # Triggers default country
    {"id": "cust004", "name": " fatima abubakar", "country": "kenya", "spent": 150000},
    {"id": "cust005", "name": "emmanuel okocha", "country": " south africa ", "spent": 45000}
]

# Empty list to store all processed profiles
processed_profiles = []

# Loop through raw customers and generate cleaned profilerrs for each
for cust in raw_customers:
    if cust["country"]:
        profile = generate_customer_profile(
        cust["id"],
        cust["name"],
        total_spent=cust["spent"],
        country=cust["country"]
        )
    else:
        profile = generate_customer_profile(
        cust["id"],
        cust["name"],
        total_spent=cust["spent"]
        )

    processed_profiles.append(profile)

# Printing all processed results neatly
print("==================================================")
print(f" {COMPANY_NAME.upper()} - CUSTOMER SYSTEM ")
print("==================================================")

for profile in processed_profiles:
    print(f"Company: {profile['company']}")
    print(f"Customer ID: {profile['customer_id']}")
    print(f"Name: {profile['name']}")
    print(f"Country: {profile['country']}")
    print(f"Spent: ₦{profile['total_spent']}")
    print(f"Status: {profile['status']}")
    print(f"Points: {profile['loyalty_points']}")
    print("--------------------------------------------------")
```


# Reflection Questions

### Question 1: What is the difference between a parameter and an argument in your project?
* **Parameter:** The placeholder variable defined in the function signature, such as `name` or `total_spent` in `def generate_customer_profile(customer_id, name, total_spent, country="Unknown"):`.
* **Argument:** The actual value or data passed into the function when calling it, such as `cust["name"]` or `125000`.

---

### Question 2: Where did you use positional arguments?
Positional arguments were used for `cust["id"]` and `cust["name"]` when calling `generate_customer_profile()`. They were passed first without explicitly naming the parameters, so Python matched them based strictly on their order in the function definition.

---

### Question 3: Where did you use keyword arguments?
Keyword arguments were used when passing `total_spent=cust["spent"]` and `country=cust["country"]` into `generate_customer_profile()`. Writing the parameter name followed by `=` explicitly told Python which value belonged to which parameter.

---

### Question 4: What would happen if you supplied the arguments in the wrong positional order?
If positional arguments are swapped (for example, passing `name` first and `customer_id` second), Python assigns the values to the wrong parameters. This causes data corruption—such as treating a customer's name as their ID—or results in a `TypeError` if a operation is performed on the wrong data type.

---

### Question 5: Which parameter in your project has a default value?
The `country` parameter in `generate_customer_profile()` has a default value of `"Unknown"` (`country="Unknown"`). If no country argument is provided when calling the function, Python automatically defaults to `"Unknown"`.

---

### Question 6: Identify one local variable. Why is it local?
`cleaned_name` (or `status`) inside `generate_customer_profile()`. It is a local variable because it was created inside the function's execution block. It only exists while the function runs and cannot be accessed outside the function in the global scope.

---

### Question 7: Identify one global variable. Why is it global?
`COMPANY_NAME` = `"SmartBiz Logistics"`. It is global because it was declared at the top level of the script outside of any function. As a result, it can be accessed anywhere in the notebook, including inside functions.

---

### Question 8: Why is using separate functions better than writing the entire program in one large block?
Separate functions break complex tasks into small, modular steps. This makes the code easier to read, debug, test, and reuse without duplicating logic across the project.

---

### Question 9: Which function in your project is responsible for cleaning data?
`clean_name()` and `clean_country()` are dedicated functions responsible for stripping whitespace and capitalizing string formatting.

---

### Question 10: Which function is responsible for business logic?
`calculate_status()` (which determines customer tiers like Gold, Silver, and Bronze based on spend) and `calculate_loyalty_points()` (which calculates points based on spending thresholds).
